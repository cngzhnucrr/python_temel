# Değişkenler

* Integer (int) Tam Sayılar — Virgülsüz
* Float (float) Kesirli Sayılar — Virgüllü
* Boolean (bool) True ya da False
* String (str) “ “ veya ‘ ‘ arasına yazılarak gösterilen karakterler bütünü

---------------------------------------------------------------------------------------------------

* type()
* type() komutu değişkenin türünü görmemizi sağlar.

---------------------------------------------------------------------------------------------------

```python
a:int=0       # tamsayı türünde tanımlanmış bir değişken 
b:str="a"     # metin türünde tanımlanmış bir değişken 
c:float=0.1   # ondalık sayı türünde tanımlanmış bir değişken 
d:bool=True   # mantıksal türünde tanımlanmış bir değişken 
```

```python
print(f"a={a} değeri {type(a)} tipindedir")
print(f"b={b} değeri {type(b)} tipindedir")
print(f"c={c} değeri {type(c)} tipindedir")
print(f"d={d} değeri {type(d)} tipindedir")
```

---------------------------------------------------------------------------------------------------

* Yerel Değişken: Nereden tanımlarsak oradan erişebildiğimiz, methoda özgü değişkenlerdir.
* Global Değişken: Yazdığımız kodun herhangi bir yerinden erişebildiğimiz değişkenlerdir.
