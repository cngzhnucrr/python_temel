# aracın servis zamanı hesaplama 

```python
import datetime 
```

```python
tarih=input("araç hangi tarihte tafiğe çıktı (1111/11/11) :")
tarih=tarih.split("/") #tarih bilgisini gün ay yıl olarak birbirinden ayırdık
print("yıl:",tarih[0])
print("ay:",tarih[1])
print("gün:",tarih[2])
trafige_cikis=datetime.datetime(int(tarih[0]),int(tarih[1]),int(tarih[2]))
simdi=datetime.datetime.now() # bu günün bilgisi alınır 
fark=simdi-trafige_cikis
gün=fark.days
print("gün",gün)
if gün<365:
    print("1. servis aralığı")
elif gün>365 and gün<=2*365:
    print("2. servis aralığı")
elif gün>2*365 and gün<=3*365:
    print("3. servis aralığı")
else:
    print("4. servis aralığı")

    
```
