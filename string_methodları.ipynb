```python
message=" Hello There. My name is Cengizhan Uçar "
```

upper( ) metodu tüm kelimeyi büyük yazar 

```python
print(message.upper())
```

lower() metodu tüm kelimeyi büyük yapar

```python
print(message.lower())
```

title( ) metodu kelimelerin baş harflerini büyük yapar

```python
print(message.title())
```

capitalize( ) metodu verilen string ifadenin sadec baş harfini büyük yapar geri kalan tüm harfleri küçük yapar

```python
print(message.capitalize())
```

strip( ) metodu string kelimenin başındaki boşluk karakterini siler

```python
print(message.strip())
```

split() methodu string bir ifadeyi indexli elemanlara ayırarak liste türüne dönderir

```python
y=message.split()
print(type(y))
print(y)
```

join( ) methodu  parça parça olan kelimeleri birleştirir

```python
y=' '.join(y)
print(y)
```

```python
" aynı methodu bir kez daha çalıştırırsak"
y=' '.join(y)
print(y)
```

find( ) metodu string içinde aramak istediğimiz kelimenin var olup olmadığını söyler bize, eğer varsa kelimenin ilk harfinin index numarasını bize döndürür, eğew -1 döndürürse kelime yok demektir.

stratswith( ) methodu ise stringin ne ile başlayıp başlamadığını kontrol eder

```python
x=message.find("is")
y=message.startswith('')
print(x)
print(y)
```

```python
print(message.startswith("i"))
```

```python
print(message.endswith("i"))
```

```python
print(message.endswith(" "))
```

```python
print(message)
```

replace( )  metodu ise içindeki ilk parametreyi bulur ve yerine ikinci parametreyi yazar

```python
print(message.replace('Cengizhan','uçç'))
```

```python
print(message.replace(' ',' bunlar_boşluk '))
```

center( ) metodu ise kullanıldığında kelimenin sağında ve solunda ne kadar boşluk yada karakter olacağını belirler

```python
print(message.center(100))
```

```python
print(message.center(50))
```

```python
print(message.center(10))
```
