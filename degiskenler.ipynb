kullanıcıya aldığı maaşı sorup %27 lik bir vergi ile aldığı maaşı hesaplayalım 


```python
maas=int(input("lütfen maaşınızı giriniz:"))
net_maas= maas- maas*0.27
print("net maaşınız:",net_maas)
```

not: input fonksiyonu aldığı verileri string veri tipinde alır

aynı fonksiyonu daha güzel bir halde kullanalım

```python
maas=int(input("lütfen maaşınızı giriniz:"))
vergi=0.27
net_maas= maas- maas*vergi
print("net maaşınız:",net_maas)
```
