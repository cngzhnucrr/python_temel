# Sözlükler

* Sözlükler, veri değerlerini anahtar:değer çiftleri halinde saklamak için kullanılır.
* Sözlük, sıralı, değiştirilebilir ve yinelenen öğelere izin vermeyen bir koleksiyondur.
* (Python 3.7 sürümünden itibaren sözlükler sıralıdır . Python 3.6 ve önceki sürümlerde sözlükler sıralı değildir)
* Sözlük öğeleri sıralıdır, değiştirilebilir ve yinelenen öğelere izin vermez.
* Sözlük öğeleri anahtar:değer çiftleri halinde sunulur ve anahtar adı kullanılarak erişilebilir.
* Sözlükler değiştirilebilir özelliktedir; yani sözlük oluşturulduktan sonra öğeleri değiştirebilir, ekleyebilir veya kaldırabiliriz.
* Sözlüklerde aynı anahtara sahip iki öğe bulunamaz
* Ayrıca `dict()` kurucu fonksiyonunu kullanarak da sözlük oluşturmak mümkündür .

```python
# Sözlükler kıvrımlı parantezlerle yazılır ve anahtarları ile değerleri vardır:
sözlük={
    "dil sürümü":3.7,
    "dil":"python",
    "yil":2026
        }
print(sözlük)
```

```python
print(sözlük["dil sürümü"])
```

```python
print(type(["dil sürümü"]))
```

```python
# Bir sözlüğün kaç öğe içerdiğini belirlemek için
print(len(sözlük))
```

```python
py_sözlük={
    "dil sürümü":3.7,
    "dil":"python",
    "yil":2026,
    "modüller":["pandas","numpy","matplotlib"]
        }
print(py_sözlük)
```

Bir sözlüğün öğelerine, köşeli parantez içinde yer alan anahtar adını kullanarak erişebilirsiniz:

```python
print(py_sözlük)
```

```python
print(py_sözlük["modüller"])
```

```python
print(py_sözlük["dil sürümü"])
```

Aynı sonucu verecek başka bir yöntem de var : get() fonksiyonu

```python
print(py_sözlük.get("modüller"))
```

* Anahtarları Al: sözlükteki tüm anahtarların bir listesini döndüren komut "keys()"
* Anahtar listesi, sözlüğün bir görünümüdür ; yani sözlükte yapılan herhangi bir değişiklik anahtar listesine de yansıyacaktır.

```python
print("py_sözlük anahtarları:",py_sözlük.keys())
```

Orijinal sözlüğe yeni bir öğe ekleyelim ve anahtar listesinin de güncelleyelim

```python
py_sözlük["kullanılan ide"]="anaconda spyder"
```

```python
print(py_sözlük.keys())
```

```python
print(py_sözlük)
```

* Değerleri alalım: "" values()""  bu yöntem, sözlükteki tüm değerlerin bir listesini döndürecektir.
* Değerler listesi, sözlüğün bir görünümüdür ; yani sözlükte yapılan herhangi bir değişiklik, değerler listesine de yansıyacaktır.

```python
print(py_sözlük.values())
```

* sözlüğün tamamını görelim: """items()""" yöntemi , sözlükteki her öğeyi bir liste halinde demetler olarak döndürecektir.
* Döndürülen liste, sözlüğün öğelerinin bir görünümüdür ; yani sözlükte yapılan herhangi bir değişiklik, öğe listesine de yansıyacaktır.

```python
print(py_sözlük.items())
```

```python
if "yil" in py_sözlük:
    print("yil sözcüğü py_sözlük içerisinde anahtar olarak kullanılmaktadır")
else:
     print("yil sözcüğü py_sözlük içerisinde anahtar olarak kullanılmamaktadır")
```

sözlüğü güncelleyelim

```python
urun={"marka":"apple","model":"iphone11","fiyat":"100 tl","yil":2020}
```

```python
print(urun)
```

```python
#önce sözlükteki anahtarları(key) alalım
print(urun.keys())
```

```python
#şimdi anahtar değerlerini öğrenelim
print(urun.get("marka"))
print(urun.get("model"))
print(urun.get("fiyat"))
print(urun.get("yil"))
```

```python
# anahtar değerlerini(values) direkt görelim
print(urun.values())
```

```python
# sözlük içindeki bir anahtarın değerini güncelleyelim
print(urun.keys())
# fiyat 300 tl olsun
urun["fiyat"]="300 tl"
print(urun.values())
print(urun)
```

```python
# şimdi (key,value) çiftini liste olarak dönürelim
print(urun.items())
```

```python
# urun sözlüğünde üreyim yılı "yil" anahtarı ile mevcut mu bunu kontrol edelim
if "yil" in urun:
    print("evet üretim yılı mevcut")
else:
    print("hayır üretim yılı mevcut değil")
```

```python
# urun sözlüğünde model bilgisinin "model" anahtarı ile tutulup tutulmadığını kontrol edelim
if "model" in urun:
    print("model bilgisi mevcut")
else:
    print("model bilgisi mevcut değil")
    
```

```python
# anahtarı kullanarak o ögenin değerini öğrenebiliriz
print(urun.keys())
```

```python
print(" yıl bilgisi:",urun["yil"])
print(" model bilgisi:",urun["model"])
```

```python
# dell markalı bilgisayarları ürünleri method kullanarak sözlük oluşturma
urunler=dict(marka="dell",model="inspiron",serii="n5110",üretim_yılı=2025)
```

```python
print(urunler)
```

```python
# sözlükteki anahtar kelimeleri  tek tek yazdıralım
for key in urunler:
    print(key)
```

```python
# sözlükteki değerleri tek tek döndürelim
for value in urunler.values():
    print(value)
```

```python
# anahtar kelimeleri keys() methodu ile de dönderebiliriz
for x in urunler.keys():
    print(x)
```

```python
# hem anahtar hem de değer ögelerini döndürelim
for x,y in urunler.items():
    print(x,y)
```

```python
print(f"en sonuncu ögenin tipine bakalım, key:{type(x)} , key:{type(y)}")
```

```python
# urunler listesinin kopyasını oluşturalım
copy=dict(urunler)
```

```python
print(copy)
```

```python
copy2=urunler.copy()
```

```python
print(copy2)
```
