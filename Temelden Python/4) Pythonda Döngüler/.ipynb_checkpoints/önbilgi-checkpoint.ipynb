# döngüler 

* Python'da döngüler, belirli bir kod bloğunu tekrar tekrar çalıştırmak için kullanılır.

* " For Döngüsü " (Eleman Tabanlı)Bir listenin, metnin (string) veya bir sayı aralığının üzerinden sırayla geçmek (iterasyon yapmak) için kullanılır. Eleman sayısı genellikle baştan bellidir.

```python
# Bir listenin elemanlarını yazdırma
meyveler = ["elma", "muz", "çilek"]
for meyve in meyveler:
    print(meyve)
```

* " range() " FonksiyonuBelirli bir sayıda dönmek için range() fonksiyonu kullanılır. Sıfırdan başlar ve belirtilen sayıya kadar (o sayı hariç) gider.

```python
# 0'dan 4'e kadar olan sayıları yazdırır (5 hariç)
for x in range(5):
    print(x)
```

* " While Döngüsü " (Koşul Tabanlı)Belirli bir koşul doğru (True) olduğu sürece çalışmaya devam eder. Koşul yanlış (False) olduğunda döngü biter.

```python
# i sayısı 6'dan küçük olduğu sürece çalışır
i = 1
while i < 6:
    print(i)
    i += 1  # Önemli: i'yi artırmazsanız döngü sonsuza kadar sürer!
```

* Döngü Kontrol İfadeleri Döngünün normal akışını değiştirmek için şu komutlar kullanılır:
*  "break": Döngüyü tamamen sonlandırır ve döngüden çıkar.
*  "continue": Döngünün o anki turunu atlar ve hemen bir sonraki tura geçer.

```python
# break örneği
for x in range(10):
    if x == 5:
        break  # x 5 olduğunda döngü tamamen biter
    print(x)   # 0, 1, 2, 3, 4 yazdırır

# continue örneği
for x in range(5):
    if x == 3:
        continue  # 3'ü atlar, yazdırmadan sonraki sayıya geçer
    print(x)      # 0, 1, 2, 4 yazdırır
```
