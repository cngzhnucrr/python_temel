# f string özelliği 

```python
olcum=176
print(f"boyunuz = {olcum} cm ve yaşınız = {yas} dır")
```

# format parametresi 

```python
print("boyunuz= {} cm ve yaşınız ={} dır".format(olcum,yas))
```

```python
print('boyunuz= {} cm ve yaşınız ={} dır'.format(olcum,yas))
```

```python
print('boyunuz= {0} cm ve yaşınız ={1} dır'.format(olcum,yas))
```

```python
print('boyunuz= {1} cm ve yaşınız ={0} dır'.format(olcum,yas))
```

format içindeki ifadeler indexlenir yukarıdada görüldüğü gibi

```python
print('boyunuz= {n} cm ve yaşınız ={s} dır'.format(n=olcum,s=yas))
```

```python
result=200/700
```

```python
print('the result is {}'.format(result))
```

```python
print('the result is {r}'.format(r=result))
```

```python
print('the result is {r:1.3}'.format(r=result))
```

```python
print('the result is {r:.3}'.format(r=result))
```

```python
print('the result is {r:.5}'.format(r=result))
```

```python
"soru 4 cevabı"
print(f"course kelimesinin uzunluğu : {len(course)}")
```

```python
word_1=course[:15]
print(f"course kelimesinin ilk 15 elemanı:{len(word_1)} adet ")
word_2=course[len(course)-15:len(course)]
print(f"course kelimesinin son 15 elemanı:{len(word_2)} adet ")
word=word_1+word_2
print("website kelimesinin ilk 15 elemanı ve son 15 elemanı = ",word)
```
