* upper()Tüm karakterleri büyük harfe dönüştürür.

```python
temiz_bilgiler=bilgiler.split(",")
```

```python
print("temiz bilgiler: ",temiz_bilgiler)
```

```python
tanimlar=["mahalle:","sokak","no:","kat","şehir","belde"]
```

```python
 bilgiler_dict={}
```

```python
n:int=0
while n<len(tanimlar):
    bilgiler_dict[tanimlar[n]]=temiz_bilgiler[n]
    n+=1
```

```python
print(bilgiler_dict)
```

```python
def kullanici_getir():
    print(website+"\n")
    print(isim+"\n")
    print(soyad+"\n")
    print(str(Tc)+"\n")
    for i in bilgiler_dict:
        print(i +"\n")
```
