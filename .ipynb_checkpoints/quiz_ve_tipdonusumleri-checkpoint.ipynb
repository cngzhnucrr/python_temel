bir müsterininib adres bilgisi dahil bilgileri için değişkenleri oluşturunuz

```python
print("musteri bilgileri ")
must_ad="Cengizhan" 
must_soyad="ucar\n"
must_cinsiyet="erkek\n"
must_tc="8732874832"
must_adres="şurası burası"
must_yas=20

must_bilgileri= "ad-soyad: "+must_ad +" "+ must_soyad+" "+"cinsiyet: "+must_cinsiyet+" "+"tc: "+ must_tc+" "+"adres: "+must_adres+" "+"yas:"+ str(must_yas)
print(must_bilgileri)
```

```python
print("müşteri bilgileri uzunluğu: ",len(must_bilgileri))
```

```python
print("müşteri bilgilerinde bulunan en son karakterin en son harfi: ",must_bilgileri[86])
```

```python
print("müşteri bilgilerinde bulunan en son karakterin en son harfi: ",must_bilgileri[-1])
```

```python
print("müşteri bilgilerinin 5. harfi: ",must_bilgileri[5])
```

```python
print("müşteri bilgilerinin 5 ve 5 den sonraki elemanları: ",must_bilgileri[5:])
```

```python
print("müşteri bilgilerinin 20 ve 20 den sonraki elemanları: ",must_bilgileri[20:])
```

# nottt

must_bilgileri değişkeni için parametreler şu şekildedir 
* * must_bilgileri[x:y:z]

* x: başlangıç index değeri
* y: bitiş index değeri
* z: artış yada azalış index değeri

# tip dönüşümleri

```python
x=4
y=4.0
z="aga"
isOnline= True
```

```python
print("x değişkeninin tipi: ", type(x))
print("y değişkeninin tipi: ", type(y))
print("z değişkeninin tipi: ", type(z))
print("isOnline değişkeninin tipi: ", type(isOnline))
```

```python
yeni_x=float(x)
yeni_isOnline=int(isOnline)
print(yeni_x)
print(yeni_isOnline)

print("x değişkeninin tipi: ", type(x))
print("isOnline değişkeninin tipi: ", type(isOnline))
print("yeni_x değişkeninin tipi: ", type(yeni_x))
print("yeni_isOnline değişkeninin tipi: ", type(yeni_isOnline))
```

# quiz: kullanıcıdan girdi alarak dairenin alanını ve çevresini hesaplayınız (pi=3.14)

```python
pi=3.14
yaricap=float(input("dairenin yarıçapını m cinsinden giriniz:"))
cevre=2*pi*yaricap
alan=pi*(yaricap**2)
print(f"dairenin çevresi:{cevre} metre ")
print(f"dairenin alanı: {alan} metrekare")
```
